```python
# ==========================================
# 1. 读取你的审查数据集
# ==========================================
import pandas as pd
print("正在读取审查数据集...")
# 已经替换为最新的正确文件名
df = pd.read_csv('data.csv', encoding='gbk')


# 丢弃空行，确保数据干净
df = df.dropna(subset=['核心诉求', '审查意见原文', '核心雷区'])
```

```python
from langchain_core.documents import Document
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
# ==========================================
# 2. 数据拼装（Data Formatting）
# 将你的表格拼接成 AI 最容易理解的“上下文文档”
# ==========================================
documents = []
for index, row in df.iterrows():
    # 将输入(X)和输出(Y)拼接成一段结构化的文本
    # 这样当 AI 检索时，能同时看到“什么诉求”导致了“什么雷区”
    content = f"""
    【案卷名称】：{row['案卷名称']}
    【开发商/设计单位诉求】：{row['核心诉求']}
    【官方技术审查意见】：{row['审查意见原文']}
    【触发的核心雷区】：{row['核心雷区']}
    """
    
    # 将每一条记录封装成 LangChain 标准的 Document 对象
    # 可以把额外的字段存为 metadata，方便以后做精确过滤
    doc = Document(
        page_content=content.strip(),
        metadata={"案卷名称": row['案卷名称'], "核心雷区标签": row['核心雷区']}
    )
    documents.append(doc)

print(f"成功构建了 {len(documents)} 条高质量审查知识文档！")
```

```python
# ==========================================
# 3. 加载向量化模型 (Embedding Model)
# 这里使用对中文支持极好的开源开源模型
# ==========================================
print("正在加载文本向量化模型 (首次运行会自动下载模型权重)...")
# shibing624/text2vec-base-chinese 是非常成熟的中文开源 embedding 模型
embeddings = HuggingFaceEmbeddings(model_name="shibing624/text2vec-base-chinese")

# ==========================================
# 4. 构建并持久化向量数据库 (Vector DB)
# 使用轻量级开源向量库 Chroma
# ==========================================
print("正在将数据灌入 Chroma 向量数据库...")
persist_directory = "./chengdu_planning_db" # 数据库存放在你电脑的这个文件夹下

# 创建数据库并直接将数据存进去
vector_db = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    persist_directory=persist_directory
)

# 保存数据库
# vector_db.persist()
print(f"太棒了！知识库已成功建立并保存在 {persist_directory} 目录下！")
```

```python
import os
from openai import OpenAI
```

```python
# ==========================================
# 步骤 1：加载你之前建好的“本地经验向量库” (RAG)
# ==========================================
print("正在唤醒 AI 审查大脑并连接历史案卷数据库...")
embeddings = HuggingFaceEmbeddings(model_name="shibing624/text2vec-base-chinese")
vector_db = Chroma(persist_directory="./chengdu_planning_db", embedding_function=embeddings)
```

```python
# ==========================================
# 步骤 2：配置大模型 (LLM) 接口
# 这里以目前国内极其火爆且便宜的 DeepSeek 为例，你也可以换成 Kimi/智谱/通义等
# ==========================================
# 请去对应大模型官网申请一个 API Key 替换下面的字符串
LLM_API_KEY = "${DEEPSEEK_API_KEY}" 
LLM_BASE_URL = "https://api.deepseek.com" # 这里是 DeepSeek 的接口，不同模型接口不同

client = OpenAI(api_key=LLM_API_KEY, base_url=LLM_BASE_URL)
```

```python
def generate_official_review(new_project_query):
    """
    核心大脑：接收新案卷，检索历史，调用大模型生成报告
    """
    print(f"\n👉 收到新案卷诉求：{new_project_query}")
    print("🔍 正在检索成都市历史相似案卷规则...")
    
    # 1. 检索最相似的 3 个历史案卷
    similar_docs = vector_db.similarity_search(new_project_query, k=3)
    
    # 2. 把历史案卷拼装成“参考上下文”
    context = ""
    for i, doc in enumerate(similar_docs):
        context += f"\n【历史参考案卷 {i+1}】\n{doc.page_content}\n"
    
    print("💡 检索完毕，正在呼叫大模型撰写官方审查报告...\n")
    
    # 3. 提示词工程 (Prompt Engineering) - 让大模型扮演审查专家
    system_prompt = """
    你现在是成都市规划和自然资源局的资深技术审查专家。
    你的任务是根据用户提供的【新案卷诉求】，结合提供的【成都市历史控规调整案例】，
    撰写一份极其专业、严谨、符合官方行文规范的《技术审查意见》。
    
    要求：
    1. 语气必须是客观、冷峻的政务官方口吻。
    2. 必须从历史案例中提取审查逻辑（如：涉轨需轨道集团意见、商改住需论证必要性、公服需满足不降低要求等）。
    3. 输出结构需包含：一、成果及内容完整性；二、核心技术审查意见；三、后续会签建议。
    """
    
    user_prompt = f"""
    【成都市历史控规调整案例参考】：
    {context}
    
    ======================
    【本次新案卷诉求】：
    {new_project_query}
    
    请根据以上历史雷区和规则，为这个新案卷出具官方审查意见。
    """

    # 4. 调用大模型生成报告
    response = client.chat.completions.create(
        model="deepseek-chat", # 替换为你使用的模型名字，如 kimi, glm-4 等
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        temperature=0.3 # 温度越低，输出越严谨客观
    )
    
    return response.choices[0].message.content
```

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
# 基础配置
LLM_API_KEY = "${DEEPSEEK_API_KEY}" 
LLM_BASE_URL = "https://api.deepseek.com/v1" 
client = OpenAI(api_key=LLM_API_KEY, base_url=LLM_BASE_URL)

print("正在唤醒 AI 审查大脑...")
embeddings = HuggingFaceEmbeddings(model_name="shibing624/text2vec-base-chinese")
vector_db = Chroma(persist_directory="./chengdu_planning_db", embedding_function=embeddings)

# ==========================================
# 动作 1：逐页阅读，记录疑点（初审员）
# ==========================================
def scan_pdf_page_by_page(pdf_path):
    print(f"\n📂 开始逐页精读案卷: {pdf_path}")
    loader = PyPDFLoader(pdf_path)
    pages = loader.load()
    
    all_findings = [] # 用来记录每一页发现的问题（初审员的小本本）
    
    # 为了演示速度，这里我们假设按每 3 页为一个 Chunk 进行阅读（防止一页页读太慢）
    chunk_size = 3
    for i in range(0, len(pages), chunk_size):
        chunk_pages = pages[i:i+chunk_size]
        chunk_text = "\n".join([p.page_content for p in chunk_pages])
        page_range = f"第 {i+1}-{min(i+chunk_size, len(pages))} 页"
        
        print(f"👀 正在审阅 {page_range}...")
        
        prompt = f"""
        你是一个细致的规划审查员，正在逐页阅读一份《控规调整方案》。
        请审阅以下几页的内容，提取其中【实质性的规划调整动作】或【潜在雷区】（如：用地性质改变、绿地/公服减少、涉及高压线/地铁、人口增加等）。
        如果这几页只是套话、目录或背景介绍，请直接回复"无"。
        如果发现实质内容，请简明扼要地列出（字数控制在100字内）。
        
        【审阅内容】：
        {chunk_text}
        """
        
        response = client.chat.completions.create(
            model="deepseek-chat",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.1
        )
        finding = response.choices[0].message.content.strip()
        
        # 如果大模型发现了问题，就记在小本本上，并标注页码
        if finding != "无" and "没有实质性" not in finding:
            all_findings.append(f"[{page_range} 发现]: {finding}")
            print(f"   🚨 记录疑点: {finding}")
            
    return "\n".join(all_findings)

# ==========================================
# 动作 2：综合疑点，检索并出具最终报告（主审处长）
# ==========================================
def generate_final_report(findings_summary):
    if not findings_summary:
        return "该案卷未发现实质性规划调整内容，请核实文件完整性。"
        
    print("\n📚 案卷精读完毕，正在去数据库检索相关历史雷区...")
    
    # 拿着整个小本本去数据库里找最相似的历史案卷
    similar_docs = vector_db.similarity_search(findings_summary, k=4)
    
    context = ""
    for idx, doc in enumerate(similar_docs):
        context += f"【参考案卷 {idx+1}】: {doc.metadata.get('案卷名称', '未知')}\n内容: {doc.page_content}\n\n"
        
    print("✍️ 正在撰写最终官方审查报告...")
    
    system_prompt = """
    你现在是成都市规划局的资深审查专家。
    刚才你的下属逐页阅读了一份案卷，并提交了【案卷疑点汇总】。
    请你结合提供的【历史审查雷区库】，出具一份官方、冷峻的《技术审查意见》。
    
    特色要求：
    1. 在指出问题时，可以引用是案卷的第几页出现的问题（根据疑点汇总的提示），显得更加严谨。
    2. 严格对标历史审查逻辑（如缺必要性论证、需市级部门会签、需查毒等）。
    3. 结构化输出：一、成果完整性；二、核心技术审查意见；三、跨部门会签建议。
    """
    
    user_prompt = f"""
    【历史审查雷区库（成都本地规矩）】：
    {context}
    
    ======================
    【下属提交的逐页案卷疑点汇总】：
    {findings_summary}
    
    请根据以上信息，撰写最终审查意见。
    """

    response = client.chat.completions.create(
        model="deepseek-chat",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        temperature=0.3
    )
    
    return response.choices[0].message.content

```

```python
if __name__ == "__main__":
    TEST_PDF_PATH = "./test/test_1.pdf" 
    
    if os.path.exists(TEST_PDF_PATH):
        # 步骤1：逐页精读
        findings = scan_pdf_page_by_page(TEST_PDF_PATH)
        
        # 步骤2：出具最终报告
        print("\n" + "="*20 + " 最终生成的官方技术审查意见 " + "="*20 + "\n")
        final_report = generate_final_report(findings)
        print(final_report)
        print("\n" + "="*68)
    else:
        print("找不到文件，请检查路径。")
```
